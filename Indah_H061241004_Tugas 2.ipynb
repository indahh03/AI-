## Eksekusi 1: Multichannel Analysis of Surface Waves (MASW)

**Data sintetis:** kurva dispersi frekuensi–kecepatan fase dan model kecepatan gelombang geser (Vs) sederhana. Data ini dibuat untuk demonstrasi, bukan hasil survei lapangan.

**Tujuan eksekusi:** memperlihatkan hubungan kurva dispersi dengan model Vs dan menghasilkan visualisasi awal untuk latihan interpretasi.


```python
import numpy as np
import matplotlib.pyplot as plt

# ============================================================
# 1. MEMBUAT KURVA DISPERSI SINTETIS
# ============================================================
frequency = np.linspace(5, 40, 100)  # Hz
# Kurva fase sintetis: kecepatan cenderung meningkat pada frekuensi rendah
phase_velocity = 180 + 420 * np.exp(-frequency / 13) + 12*np.sin(frequency/4)

# Model Vs sederhana berdasarkan kedalaman
depth = np.array([0, 5, 12, 25, 40])
vs = np.array([180, 180, 320, 520, 520])

print("MASW - DATA SINTETIS")
print(f"Rentang frekuensi       : {frequency.min():.1f}–{frequency.max():.1f} Hz")
print(f"Rentang kecepatan fase  : {phase_velocity.min():.1f}–{phase_velocity.max():.1f} m/s")
print("Model Vs lapisan        : 180 m/s (0–5 m), 320 m/s (5–12 m), 520 m/s (>12 m)")
print("Catatan: nilai dibuat untuk demonstrasi, bukan estimasi kondisi lapangan.")

fig, ax = plt.subplots(1, 2, figsize=(11, 4.5))
ax[0].plot(frequency, phase_velocity, linewidth=2)
ax[0].set_xlabel("Frekuensi (Hz)")
ax[0].set_ylabel("Kecepatan fase (m/s)")
ax[0].set_title("Kurva dispersi sintetis")
ax[0].grid(True, alpha=0.3)

ax[1].step(vs, depth, where="post", linewidth=2)
ax[1].invert_yaxis()
ax[1].set_xlabel("Vs (m/s)")
ax[1].set_ylabel("Kedalaman (m)")
ax[1].set_title("Model Vs sintetis")
ax[1].grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

**Interpretasi hasil:** Kurva dispersi sintetis menunjukkan perubahan kecepatan fase terhadap frekuensi. Model Vs memperlihatkan kecepatan 180 m/s pada lapisan dangkal, meningkat menjadi 320 m/s pada kedalaman 5–12 m, lalu 520 m/s di bawah 12 m. Kenaikan Vs dapat mengindikasikan material yang semakin kaku atau kompak, tetapi interpretasi lapangan perlu didasarkan pada proses inversi kurva dispersi, kualitas data, dan informasi geologi pendukung.

## Eksekusi 2: Horizontal-to-Vertical Spectral Ratio (HVSR)

**Data sintetis:** kurva H/V dengan satu puncak dominan pada frekuensi sekitar 2 Hz. Kurva ini hanya contoh untuk memperlihatkan cara membaca frekuensi dan amplitudo puncak.

**Tujuan eksekusi:** menghitung parameter puncak dari kurva sintetis dan memvisualisasikan respons H/V.


```python
import numpy as np
import matplotlib.pyplot as plt

# ============================================================
# 1. MEMBUAT KURVA H/V SINTETIS
# ============================================================
frequency = np.linspace(0.2, 10, 500)
peak_frequency_model = 2.0
hv = 1.4 + 3.8*np.exp(-0.5*((frequency-peak_frequency_model)/0.32)**2)
hv += 0.12*np.sin(2.2*frequency)  # variasi kecil sintetis
hv = np.maximum(hv, 0.2)

peak_idx = np.argmax(hv)
f_peak = frequency[peak_idx]
a_peak = hv[peak_idx]

print("HVSR - DATA SINTETIS")
print(f"Rentang frekuensi : {frequency.min():.2f}–{frequency.max():.2f} Hz")
print(f"Frekuensi puncak  : {f_peak:.2f} Hz")
print(f"Amplitudo H/V     : {a_peak:.2f}")
print("Catatan: parameter ini berasal dari kurva simulasi, bukan pengukuran lapangan.")

plt.figure(figsize=(9, 4.5))
plt.plot(frequency, hv, label="Kurva H/V sintetis")
plt.scatter([f_peak], [a_peak], zorder=3, label=f"Puncak: {f_peak:.2f} Hz")
plt.axvline(f_peak, linestyle="--", alpha=0.7)
plt.xlabel("Frekuensi (Hz)")
plt.ylabel("Rasio H/V")
plt.title("Kurva HVSR sintetis")
plt.grid(True, alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()
```

**Interpretasi hasil:** Kurva simulasi memiliki puncak H/V sekitar 2 Hz dengan amplitudo sekitar 5. Puncak tersebut menunjukkan frekuensi dominan pada data contoh. Frekuensi puncak HVSR dapat membantu menilai respons dinamik tapak, tetapi tidak cukup untuk menetapkan kedalaman lapisan secara langsung tanpa informasi tambahan, model kecepatan, dan evaluasi kualitas kurva.

## Eksekusi 3: Ground Penetrating Radar (GPR)

**Data sintetis:** radargram pada lintasan 0–30 m dengan jendela waktu 0–80 ns, reflektor berlapis, dan satu anomali hiperbola buatan.

**Tujuan eksekusi:** menghasilkan radargram ilustratif dan menunjukkan lokasi pola refleksi yang perlu diperiksa.


```python
import numpy as np
import matplotlib.pyplot as plt

# ============================================================
# 1. PEMODELAN RADARGRAM SINTETIS
# ============================================================
x = np.linspace(0, 30, 241)
t_ns = np.linspace(0, 80, 201)
X, T = np.meshgrid(x, t_ns)
rng = np.random.default_rng(11)
radar = 0.16*rng.normal(size=X.shape)

# Reflektor berlapis
radar += 0.65*np.exp(-((T-20)/2.3)**2)*np.cos(2*np.pi*(T-20)/5)
radar += 0.48*np.exp(-((T-43)/3.0)**2)*np.cos(2*np.pi*(T-43)/6)

# Anomali hiperbola sintetis di sekitar x = 17 m
v = 0.10  # m/ns, asumsi
x0, t0 = 17.0, 28.0
hyper_t = np.sqrt(t0**2 + (2*(X-x0)/v)**2) * 0.25
# Bentuk hiperbola dibuat sebagai respons amplitudo lokal pada radargram
radar += 0.9*np.exp(-((T-hyper_t)/1.6)**2)

print("GPR - DATA SINTETIS")
print(f"Luas lintasan         : {x.min():.1f}–{x.max():.1f} m")
print(f"Jendela waktu         : {t_ns.min():.1f}–{t_ns.max():.1f} ns")
print(f"Posisi anomali model  : sekitar {x0:.1f} m")
print("Catatan: radargram merupakan simulasi untuk demonstrasi.")

plt.figure(figsize=(10, 5))
plt.imshow(radar, extent=[x.min(), x.max(), t_ns.max(), t_ns.min()],
           aspect="auto", cmap="seismic", vmin=-1.5, vmax=1.5)
plt.colorbar(label="Amplitudo relatif")
plt.xlabel("Jarak lintasan (m)")
plt.ylabel("Waktu tempuh dua arah (ns)")
plt.title("Radargram GPR sintetis")
plt.tight_layout()
plt.show()
```

**Interpretasi hasil:** Radargram memperlihatkan reflektor yang relatif mendatar pada waktu sekitar 20 ns dan 43 ns, serta pola melengkung di sekitar posisi lintasan 17 m. Reflektor mendatar dapat berkaitan dengan batas material, sedangkan pola hiperbola dapat menunjukkan objek lokal atau perubahan kondisi bawah permukaan. Bentuk tersebut tidak dapat memastikan jenis objek tanpa kalibrasi kecepatan, data lapangan, dan informasi pendukung.

## Eksekusi 4: Differential Interferometric Synthetic Aperture Radar (DInSAR)

**Data sintetis:** peta displacement Line of Sight (LOS) pada area grid 0–1000 m dengan satu zona perubahan positif dan satu zona perubahan negatif.

**Tujuan eksekusi:** memvisualisasikan pola spasial displacement dan menghitung rentang nilai pada model simulasi.


```python
import numpy as np
import matplotlib.pyplot as plt

# ============================================================
# 1. MEMBUAT PETA DISPLACEMENT LOS SINTETIS
# ============================================================
x = np.linspace(0, 1000, 101)
y = np.linspace(0, 1000, 101)
X, Y = np.meshgrid(x, y)

# Dua komponen anomali sintetis; satu positif dan satu negatif
positive = 18*np.exp(-(((X-350)/180)**2 + ((Y-600)/220)**2)/2)
negative = -12*np.exp(-(((X-700)/170)**2 + ((Y-350)/190)**2)/2)
displacement = positive + negative

print("DInSAR - DATA SINTETIS")
print(f"Displacement LOS minimum : {displacement.min():.2f} mm")
print(f"Displacement LOS maksimum: {displacement.max():.2f} mm")
print("Area model               : 0–1000 m pada sumbu X dan Y")
print("Catatan: tanda positif/negatif mengikuti konvensi model ini; bukan data deformasi aktual.")

plt.figure(figsize=(8, 6))
im = plt.imshow(displacement, extent=[x.min(), x.max(), y.min(), y.max()],
                origin="lower", cmap="RdBu_r", vmin=-18, vmax=18)
plt.colorbar(im, label="Displacement LOS (mm)")
plt.xlabel("X lokal (m)")
plt.ylabel("Y lokal (m)")
plt.title("Peta displacement LOS DInSAR sintetis")
plt.tight_layout()
plt.show()
```

**Interpretasi hasil:** Peta simulasi menunjukkan zona displacement positif dengan nilai maksimum di sekitar bagian barat-tengah area dan zona displacement negatif di bagian tenggara. Pola ini menggambarkan bahwa perubahan LOS dapat bervariasi secara spasial. Tanda positif atau negatif tidak boleh langsung disamakan dengan uplift atau subsidence tanpa mengetahui konvensi tanda produk, arah pandang satelit, geometri orbit, dan komponen gerak permukaan.

**Data sintetis:** penampang resistivitas 2D sepanjang 100 m hingga kedalaman 20 m dengan zona resistivitas rendah di bagian tengah.

**Tujuan eksekusi:** menampilkan variasi resistivitas lateral dan vertikal sebagai contoh dasar interpretasi penampang geolistrik.


```python
import numpy as np
import matplotlib.pyplot as plt

# ============================================================
# 1. MEMBUAT PENAMPANG RESISTIVITAS SINTETIS
# ============================================================
x = np.linspace(0, 100, 101)
z = np.linspace(0, 20, 61)
X, Z = np.meshgrid(x, z)

rho = np.full_like(X, 100.0)  # latar belakang ohm.m
target = ((X >= 35) & (X <= 65) & (Z >= 4) & (Z <= 13))
rho[target] = 25.0

print("GEOLISTRIK RESISTIVITAS - DATA SINTETIS")
print(f"Rentang resistivitas : {rho.min():.1f}–{rho.max():.1f} ohm.m")
print("Zona resistivitas rendah: X = 35–65 m; kedalaman = 4–13 m")
print("Catatan: nilai model dibuat untuk demonstrasi, bukan hasil inversi lapangan.")

plt.figure(figsize=(10, 5))
im = plt.imshow(rho, extent=[x.min(), x.max(), z.max(), z.min()],
                aspect="auto", cmap="viridis", vmin=20, vmax=110)
plt.colorbar(im, label="Resistivitas (ohm.m)")
plt.xlabel("Jarak lintasan (m)")
plt.ylabel("Kedalaman (m)")
plt.title("Penampang resistivitas 2D sintetis")
plt.tight_layout()
plt.show()
```

**Interpretasi hasil:** Penampang model menunjukkan zona resistivitas rendah sebesar 25 ohm.m pada jarak 35–65 m dan kedalaman 4–13 m, sedangkan latar belakang bernilai 100 ohm.m. Zona resistivitas rendah dapat berkaitan dengan material lebih konduktif, misalnya lempung atau material dengan kandungan air lebih tinggi. Namun, nilai resistivitas tidak bersifat unik sehingga interpretasi perlu mempertimbangkan kondisi geologi, hidrogeologi, konfigurasi elektroda, dan hasil inversi.
